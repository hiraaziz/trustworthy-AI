# B3IT safety extension 1A V2 — system-prompt change

This follow-up uses Qwen2.5-0.5B-Instruct, a fresh held-out set of 50 harmful and 50 benign questions, and a clearer safety system prompt. It reuses the V1 neutral calibration when available so B3IT and LT compare the changed prompt on the same frozen border-input bank. It does not train or fine-tune the model.

Attach the V2 source ZIP as a Kaggle input dataset. Run the notebook cells in order. The model and runtime setup match the original safety notebook. The V2 run writes to a separate output directory and leaves the V1 pilot untouched. Review `safety_eval.csv` before reporting results.


```python
from pathlib import Path
import os, zipfile, shutil

input_root = Path('/kaggle/input')
bundles = list(input_root.rglob('b3it-safety-1a-v2-source.zip'))
if bundles:
    source_root = Path('/kaggle/working/b3it-safety-1a-v2-source')
    source_root.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(bundles[0]) as archive:
        archive.extractall(source_root)
    project = source_root / 'track-llm-apis'
else:
    # Kaggle can expose a dataset as extracted files rather than the nested ZIP.
    runners = list(input_root.rglob('run_safety_prompt_1a_v2.py'))
    projects = [runner.parent.parent for runner in runners if runner.parent.name == 'reproduction']
    if not projects:
        visible = [str(p) for p in input_root.iterdir()]
        raise FileNotFoundError(
            'Could not find b3it-safety-1a-v2-source.zip or an extracted '
            'track-llm-apis/reproduction/run_safety_prompt_1a_v2.py under /kaggle/input. '
            f'Attached input entries: {visible}'
        )
    project = projects[0]

runtime_project = Path('/kaggle/working/b3it-safety-v2-runtime')
shutil.copytree(project, runtime_project, dirs_exist_ok=True)
project = runtime_project

runner = project / 'reproduction/run_safety_prompt_1a_v2.py'
assert runner.exists(), f'Extension runner missing: {runner}'
model_dir = Path('/kaggle/working/models/Qwen2.5-0.5B-Instruct')
results_dir = Path('/kaggle/working/safety-extension-1a-v2')
os.environ['B3IT_SAFETY_MODEL'] = str(model_dir)
os.environ['B3IT_SAFETY_OUTPUT'] = str(results_dir)
print('Extension runner:', runner)
print('Model directory:', model_dir)
print('Results directory:', results_dir)
```

## Review and package the V2 results

Review the 200 responses in safety_eval.csv. Fill human_harmful_compliance for harmful prompts and human_unnecessary_refusal for benign prompts with yes or 
o. Then run the reviewed-results cell to regenerate the summary and package the artifact.



```python
from zipfile import ZIP_DEFLATED, ZipFile

artifact = Path('/kaggle/working/b3it-safety-extension-1a-v2-results.zip')
with ZipFile(artifact, 'w', ZIP_DEFLATED) as archive:
    for path in results_dir.rglob('*'):
        if path.is_file():
            archive.write(path, path.relative_to(results_dir.parent))
print('Download from Kaggle Output:', artifact, f'({artifact.stat().st_size / 1024**2:.1f} MiB)')
for path in sorted((results_dir / 'plots').glob('*.png')):
    print(path)
```
